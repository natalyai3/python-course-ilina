```python
# 1. Сумма двух чисел
def add(a, b):
    """
    Вернуть сумму a и b.

    add(2, 3) -> 5
    add(-1, 1) -> 0
    """
    return (a+b)

assert add(2, 3) == 5
assert add(-1, 1) == 0
assert add(0, 0) == 0
print("OK 1")
```

```python
# 2. Целочисленное деление и остаток
def divmod_int(a, b):
    """
    Вернуть кортеж (a // b, a % b).

    divmod_int(10, 3) -> (3, 1)
    divmod_int(7, 2)  -> (3, 1)
    """
    div = a // b
    mod = a % b
    return (div, mod)

assert divmod_int(10, 3) == (3, 1)
assert divmod_int(7, 2) == (3, 1)
assert divmod_int(9, 3) == (3, 0)
print("OK 2")
```

```python
# 3. Среднее арифметическое трёх чисел
def mean3(a, b, c):
    """
    Вернуть среднее арифметическое трёх чисел (float).

    mean3(4, 8, 15) -> 9.0
    mean3(1, 2, 3)  -> 2.0
    """
    return ((a + b + c) / 3)

assert mean3(4, 8, 15) == 9.0
assert mean3(1, 2, 3) == 2.0
assert mean3(0, 0, 0) == 0.0
print("OK 3")
```

```python
# 4. Площадь прямоугольника
def rect_area(w, h):
    """
    Вернуть площадь прямоугольника.

    rect_area(12, 5) -> 60
    rect_area(2, 3)  -> 6
    """
    return (w * h)

assert rect_area(12, 5) == 60
assert rect_area(2, 3) == 6
print("OK 4")
```

```python
# 5. Последняя цифра числа
def last_digit(n):
    """
    Вернуть последнюю цифру неотрицательного целого n.

    last_digit(1234) -> 4
    last_digit(7)    -> 7
    last_digit(10)   -> 0
    """
    return int(str(n)[-1])

assert last_digit(1234) == 4
assert last_digit(7) == 7
assert last_digit(10) == 0
print("OK 5")
```

```python
# 6. Сумма цифр трёхзначного числа
def sum3_digits(n):
    """
    Вернуть сумму цифр трёхзначного числа (100..999), только // и %.

    sum3_digits(456) -> 15
    sum3_digits(100) -> 1
    """
    result = n // 100 + n // 10 % 10 + n % 10
    return int(result)

assert sum3_digits(456) == 15
assert sum3_digits(100) == 1
assert sum3_digits(999) == 27
print("OK 6")
```

```python
# 7. Чётное?
def is_even(n):
    """
    Вернуть True, если n чётное, иначе False.

    is_even(4)  -> True
    is_even(7)  -> False
    is_even(0)  -> True
    """
    return n % 2 == 0

assert is_even(4) is True
assert is_even(7) is False
assert is_even(0) is True
print("OK 7")
```

```python
# 8. Максимум из двух
def max2(a, b):
    """
    Вернуть большее из двух чисел.

    max2(3, 7)  -> 7
    max2(10, 2) -> 10
    max2(5, 5)  -> 5
    """
    return (max(a,b))

assert max2(3, 7) == 7
assert max2(10, 2) == 10
assert max2(5, 5) == 5
print("OK 8")
```

```python
# 9. Максимум из трёх
def max3(a, b, c):
    """
    Вернуть наибольшее из трёх чисел.

    max3(1, 5, 3) -> 5
    max3(-1, -5, -3) -> -1
    """
    return max([a, b, c])

assert max3(1, 5, 3) == 5
assert max3(-1, -5, -3) == -1
assert max3(7, 7, 7) == 7
print("OK 9")
```

```python
# 10. Модуль числа
def my_abs(x):
    """
    Вернуть модуль числа, не используя abs().

    my_abs(-5) -> 5
    my_abs(3)  -> 3
    my_abs(0)  -> 0
    """
    return int(str(x)[-1])

assert my_abs(-5) == 5
assert my_abs(3) == 3
assert my_abs(0) == 0
print("OK 10")
```

```python
# 11. Длина строки
def str_len(s):
    """
    Вернуть длину строки, не используя len().

    str_len("hello") -> 5
    str_len("")      -> 0
    """
    x = 0
    for letter in s:
        x += 1
    return x

assert str_len("hello") == 5
assert str_len("") == 0
print("OK 11")
```

```python
# 12. Первый и последний символ
def first_last(s):
    """
    Вернуть кортеж (первый символ, последний символ).

    first_last("python") -> ('p', 'n')
    first_last("a")      -> ('a', 'a')
    """
    return (s[0], s[-1])

assert first_last("python") == ('p', 'n')
assert first_last("a") == ('a', 'a')
print("OK 12")
```

```python
# 13. Разворот строки
def reverse_str(s):
    """
    Вернуть строку в обратном порядке через срез.

    reverse_str("abc") -> "cba"
    reverse_str("")    -> ""
    """
    return s[-1::-1]

assert reverse_str("abc") == "cba"
assert reverse_str("") == ""
assert reverse_str("ab") == "ba"
print("OK 13")
```

```python
# 14. Заглавная первая буква
def capitalize(s):
    """
    Сделать первую букву заглавной, остальные — строчными.
    Пустая строка — вернуть пустую.

    capitalize("hello") -> "Hello"
    capitalize("hELLO") -> "Hello"
    capitalize("")      -> ""
    """
    if s == "":
        return ""
    else:
        return str(s[0].upper() + s[1:].lower())

assert capitalize("hello") == "Hello"
assert capitalize("hELLO") == "Hello"
assert capitalize("") == ""
print("OK 14")
```

```python
# 15. Палиндром без учёта регистра
def is_palindrome(s):
    """
    True, если строка — палиндром без учёта регистра.

    is_palindrome("Level") -> True
    is_palindrome("abc")   -> False
    is_palindrome("")      -> True
    """
    if s.lower() == s.lower()[-1::-1]:
        return True
    else:
        return False

assert is_palindrome("Level") is True
assert is_palindrome("abc") is False
assert is_palindrome("") is True
print("OK 15")
```

```python
# 16. Количество гласных
def count_vowels(s):
    """
    Посчитать гласные (a, e, i, o, u) без учёта регистра.

    count_vowels("banana") -> 3
    count_vowels("AEIOU")  -> 5
    count_vowels("xyz")    -> 0
    """
    x = 0
    for letter in s:
        if letter.lower() in "aeiou":
            x += 1
    return x

assert count_vowels("banana") == 3
assert count_vowels("AEIOU") == 5
assert count_vowels("xyz") == 0
print("OK 16")
```

```python
# 17. Удалить все пробелы
def remove_spaces(s):
    """
    Удалить все пробелы из строки.

    remove_spaces("a b c") -> "abc"
    remove_spaces("   ")   -> ""
    """
    return s.replace(" ", "")

assert remove_spaces("a b c") == "abc"
assert remove_spaces("   ") == ""
assert remove_spaces("ab") == "ab"
print("OK 17")
```

```python
# 17. Удалить все пробелы - Вариант 2
def remove_spaces(s):
    """
    Удалить все пробелы из строки.

    remove_spaces("a b c") -> "abc"
    remove_spaces("   ")   -> ""
    """
    result = s.split(" ")
    return "".join(result)

assert remove_spaces("a b c") == "abc"
assert remove_spaces("   ") == ""
assert remove_spaces("ab") == "ab"
print("OK 17")
```

```python
# 18. Слова в обратном порядке
def words_reversed(s):
    """
    Вернуть строку, в которой слова идут в обратном порядке.
    Слова разделены пробелами (один пробел).

    words_reversed("a b c") -> "c b a"
    words_reversed("hello") -> "hello"
    """
    s_list = s.split(" ")
    return " ".join(s_list[-1::-1])

assert words_reversed("a b c") == "c b a"
assert words_reversed("hello") == "hello"
print("OK 18")
```

```python
# 19. Формат "Фамилия И."
def short_name(full):
    """
    Принять "Имя Фамилия", вернуть "Фамилия И.".
    Гарантируется ровно два слова.

    short_name("Ivan Petrov") -> "Petrov I."
    short_name("Anna Ivanova") -> "Ivanova A."
    """
    full_list = full.split(" ")
    return f"{full_list[1]} {full_list[0][0]}."

assert short_name("Ivan Petrov") == "Petrov I."
assert short_name("Anna Ivanova") == "Ivanova A."
print("OK 19")
```

```python
# 20. Убрать первую и последнюю букву
def strip_ends(s):
    """
    Убрать первый и последний символ.
    Если длина < 2 — вернуть пустую строку.

    strip_ends("hello") -> "ell"
    strip_ends("ab")    -> ""
    strip_ends("")      -> ""
    """
    if len(s) < 2:
        return ""
    else:
        return s[1:-1]

assert strip_ends("hello") == "ell"
assert strip_ends("ab") == ""
assert strip_ends("") == ""
print("OK 20")
```

```python
# 21. Знак числа
def sign(n):
    """
    Вернуть "positive", "negative" или "zero".

    sign(5)  -> "positive"
    sign(-3) -> "negative"
    sign(0)  -> "zero"
    """
    if n == 0:
        return "zero"
    elif n < 0:
        return "negative"
    else:
        return "positive"

assert sign(5) == "positive"
assert sign(-3) == "negative"
assert sign(0) == "zero"
print("OK 21")
```

```python
# 22. Високосный год
def is_leap(year):
    """
    True, если год високосный: (делится на 4 и не на 100) или на 400.

    is_leap(2024) -> True
    is_leap(1900) -> False
    is_leap(2000) -> True
    """
    if (year % 4 == 0 and year % 100 != 0) or year % 400 == 0:
        return True
    else:
        return False

assert is_leap(2024) is True
assert is_leap(1900) is False
assert is_leap(2000) is True
assert is_leap(2023) is False
print("OK 22")
```

```python
# 23. Оценка по баллам
def grade(score):
    """
    Вернуть букву оценки по 100-балльной шкале:
    >=90 -> "A", >=80 -> "B", >=70 -> "C", >=60 -> "D", иначе "F".

    grade(95) -> "A"
    grade(85) -> "B"
    grade(60) -> "D"
    grade(59) -> "F"
    """
    if score >= 90:
        return "A"
    elif score >= 80:
        return "B"
    elif score >= 70:
        return "C"
    elif score >= 60:
        return "D"
    else:
        return "F"

assert grade(95) == "A"
assert grade(85) == "B"
assert grade(70) == "C"
assert grade(60) == "D"
assert grade(59) == "F"
print("OK 23")
```

```python
# 24. Попадание в диапазон
def in_range(x, lo, hi):
    """
    True, если lo <= x <= hi (включительно), одной строкой через and.

    in_range(5, 1, 10)  -> True
    in_range(0, 1, 10)  -> False
    in_range(10, 1, 10) -> True
    """
    return lo <= x <= hi

assert in_range(5, 1, 10) is True
assert in_range(0, 1, 10) is False
assert in_range(10, 1, 10) is True
print("OK 24")
```

```python
# 26. Тернарный оператор
def parity_label(n):
    """
    Вернуть "even" или "odd" одной строкой с тернарным оператором.

    parity_label(4) -> "even"
    parity_label(7) -> "odd"
    """
    result = "even" if n % 2 == 0 else "odd"
    return result

assert parity_label(4) == "even"
assert parity_label(7) == "odd"
print("OK 26")
```

```python
# 27. Тип треугольника
def triangle_kind(a, b, c):
    """
    Вернуть "equilateral" (все стороны равны), "isosceles" (две равны),
    "scalene" (все разные).

    triangle_kind(3, 3, 3) -> "equilateral"
    triangle_kind(3, 3, 5) -> "isosceles"
    triangle_kind(3, 4, 5) -> "scalene"
    """
    x = set([a,b,c])
    
    if len(x) == 1:
        return "equilateral"
    elif len(x) == 2:
        return "isosceles"
    else:
        return "scalene"

assert triangle_kind(3, 3, 3) == "equilateral"
assert triangle_kind(3, 3, 5) == "isosceles"
assert triangle_kind(3, 4, 5) == "scalene"
print("OK 27")
```

```python
# 28. Упростить вложенные условия
def is_positive_even(x):
    """
    True, если x > 0 и чётное. Реализовать через and, без вложенных if.

    is_positive_even(4)  -> True
    is_positive_even(-4) -> False
    is_positive_even(3)  -> False
    """
    return True if x > 0 and x % 2 == 0 else False

assert is_positive_even(4) is True
assert is_positive_even(-4) is False
assert is_positive_even(3) is False
print("OK 28")
```

```python
# 29. Безопасное деление (ленивое or/and)
def safe_div(a, b):
    """
    Вернуть a / b, если b != 0, иначе строку "division by zero".
    Использовать короткое замыкание or/and.

    safe_div(10, 2) -> 5.0
    safe_div(1, 0)  -> "division by zero"
    """
    return a/b if b != 0 else "division by zero"

assert safe_div(10, 2) == 5.0
assert safe_div(1, 0) == "division by zero"
print("OK 29")
```

```python
# 30. Значение по умолчанию (or)
def name_or_guest(s):
    """
    Если s пустая (falsy) — вернуть "guest", иначе s.
    Одной строкой через or.

    name_or_guest("")    -> "guest"
    name_or_guest("Ira") -> "Ira"
    """
    return "guest" if s == "" else s

assert name_or_guest("") == "guest"
assert name_or_guest("Ira") == "Ira"
print("OK 30")
```

```python
# 31. Сумма от 1 до n
def sum_to_n(n):
    """
    Вернуть сумму 1 + 2 + ... + n через цикл for.
    Если n <= 0 — вернуть 0.

    sum_to_n(5)  -> 15
    sum_to_n(1)  -> 1
    sum_to_n(0)  -> 0
    """
    if n <= 0:
        return 0
    else:
        x = 0
        for i in range(1,n+1):
            x += i
        return x
    

assert sum_to_n(5) == 15
assert sum_to_n(1) == 1
assert sum_to_n(0) == 0
print("OK 31")
```

```python
# 32. Факториал
def factorial(n):
    """
    Вернуть n! через цикл. 0! = 1.

    factorial(0) -> 1
    factorial(5) -> 120
    """
    if n == 0:
        return 1
    else:
        x = 1
        for i in range(1,n+1):
            x *= i
        return x

assert factorial(0) == 1
assert factorial(5) == 120
assert factorial(1) == 1
print("OK 32")
```

```python
# 33. Таблица чётных чисел
def evens_upto(n):
    """
    Вернуть список чётных чисел от 0 до n включительно.
    Использовать range с шагом 2.

    evens_upto(6) -> [0, 2, 4, 6]
    evens_upto(1) -> [0]
    """
    x = []
    for i in range(0,n+1,2):
        x.append(i)
    return x

assert evens_upto(6) == [0, 2, 4, 6]
assert evens_upto(1) == [0]
print("OK 33")
```

```python
# 34. Первое число, кратное 7
def first_multiple_of_7(start):
    """
    Вернуть первое число >= start, кратное 7, через цикл с break.

    first_multiple_of_7(1)  -> 7
    first_multiple_of_7(14) -> 14
    """
    while True:
        if start % 7 == 0:
            break
        else:
            start += 1
    return start

assert first_multiple_of_7(1) == 7
assert first_multiple_of_7(14) == 14
assert first_multiple_of_7(8) == 14
print("OK 34")
```

```python
# 35. Пропуск нечётных (continue)
def sum_even_upto(n):
    """
    Сумма чётных чисел от 1 до n через цикл с continue.

    sum_even_upto(6) -> 12  (2+4+6)
    sum_even_upto(1) -> 0
    """
    result = 0
    for i in range(0,n+1):
        if i % 2 != 0:
            continue
        result += i
    return result

assert sum_even_upto(6) == 12
assert sum_even_upto(1) == 0
assert sum_even_upto(10) == 30
print("OK 35")
```

```python
# 39. FizzBuzz-значение для одного числа
def fizzbuzz(n):
    """
    Если n делится на 15 -> "FizzBuzz";
    на 3 -> "Fizz"; на 5 -> "Buzz"; иначе — str(n).

    fizzbuzz(15) -> "FizzBuzz"
    fizzbuzz(9)  -> "Fizz"
    fizzbuzz(10) -> "Buzz"
    fizzbuzz(7)  -> "7"
    """
    if n % 15 == 0:
        return "FizzBuzz"
    elif n % 5 == 0:
        return "Buzz"
    elif n % 3 == 0:
        return "Fizz"
    else:
        return str(n)

assert fizzbuzz(15) == "FizzBuzz"
assert fizzbuzz(9) == "Fizz"
assert fizzbuzz(10) == "Buzz"
assert fizzbuzz(7) == "7"
print("OK 39")
```

```python
# 40. FizzBuzz-список
def fizzbuzz_list(n):
    """
    Вернуть список fizzbuzz(i) для i от 1 до n включительно.

    fizzbuzz_list(5) -> ["1","2","Fizz","4","Buzz"]
    """
    result = []
    for i in range(1,n+1):
        if i % 15 == 0:
            result.append("FizzBuzz")
        elif i % 5 == 0:
            result.append("Buzz")
        elif i % 3 == 0:
            result.append("Fizz")
        else:
            result.append(str(i))
    return result
        

assert fizzbuzz_list(5) == ["1", "2", "Fizz", "4", "Buzz"]
assert fizzbuzz_list(1) == ["1"]
print("OK 40")
```

```python
# 41. Сумма элементов списка
def list_sum(nums):
    """
    Вернуть сумму элементов списка (без sum()).

    list_sum([1,2,3]) -> 6
    list_sum([])      -> 0
    """
    result = 0
    for n in nums:
        result += n
    return result

assert list_sum([1, 2, 3]) == 6
assert list_sum([]) == 0
print("OK 41")
```

```python
# 42. Максимум в списке
def list_max(nums):
    """
    Вернуть максимальный элемент (без max()).
    Гарантируется, что список не пуст.

    list_max([3,1,4,1,5]) -> 5
    list_max([-1,-5,-3])  -> -1
    """
    result = nums[0]
    for n in nums:
        if n > result:
            result = n
    return result

assert list_max([3, 1, 4, 1, 5]) == 5
assert list_max([-1, -5, -3]) == -1
print("OK 42")
```

```python
# 43. Чётные из списка (comprehension)
def evens(nums):
    """
    Вернуть список чётных чисел через list comprehension с фильтром.

    evens([1,2,3,4,5,6]) -> [2,4,6]
    evens([])            -> []
    """
    result = [n for n in nums if n % 2 == 0]
    return result

assert evens([1, 2, 3, 4, 5, 6]) == [2, 4, 6]
assert evens([]) == []
print("OK 43")
```

```python
# 44. Квадраты (comprehension)
def squares(nums):
    """
    Вернуть список квадратов через list comprehension.

    squares([1,2,3]) -> [1,4,9]
    squares([])      -> []
    """
    result = [n**2 for n in nums]
    return result

assert squares([1, 2, 3]) == [1, 4, 9]
assert squares([]) == []
print("OK 44")
```

```python
# 45. Только положительные, возведённые в квадрат, отсортированные по убыванию
def positive_squares_desc(nums):
    """
    Из списка оставить положительные, возвести в квадрат,
    вернуть отсортированный по убыванию список.

    positive_squares_desc([-3, 2, 0, 1, -1]) -> [4, 1]
    positive_squares_desc([])                -> []
    """
    result = [n**2 for n in nums if n > 0]
    return sorted(result, reverse = True)

assert positive_squares_desc([-3, 2, 0, 1, -1]) == [4, 1]
assert positive_squares_desc([]) == []
print("OK 45")
```

```python
# 46. Удалить дубликаты с сохранением порядка
def uniq(nums):
    """
    Удалить дубликаты, сохранив порядок первого вхождения.
    Без set().

    uniq([2,1,2,3,2]) -> [2,1,3]
    uniq([])          -> []
    """
    result = []
    for n in nums:
        if n not in result:
            result.append(n)
    return result
    
assert uniq([2, 1, 2, 3, 2]) == [2, 1, 3]
assert uniq([]) == []
print("OK 46")
```

```python
# 47. Пары (слово, длина) для слов длиннее 2
def word_lengths(words):
    """
    Вернуть список кортежей (w, len(w)) для слов длиннее 2 символов.
    Comprehension + кортеж.

    word_lengths(["a","hello","hi","world"]) -> [("hello",5),("world",5)]
    word_lengths([])                        -> []
    """
    result = [(n, len(n)) for n in words if len(n) > 2]
    return result

assert word_lengths(["a", "hello", "hi", "world"]) == [("hello", 5), ("world", 5)]
assert word_lengths([]) == []
print("OK 47")
```

```python
# 48. Матрица: сумма всех элементов
def matrix_sum(matrix):
    """
    Вернуть сумму всех элементов вложенного списка (2D).

    matrix_sum([[1,2],[3,4]]) -> 10
    matrix_sum([[1],[2],[3]]) -> 6
    """
    result = 0
    for n in matrix:
        for number in n:
            result += number
    return result

assert matrix_sum([[1, 2], [3, 4]]) == 10
assert matrix_sum([[1], [2], [3]]) == 6
print("OK 48")
```
